```python
import kagglehub
import os
import pandas as pd

# Download latest version
path = kagglehub.dataset_download("harishyadav0506/youtube-shorts-dataset")

print("Path to dataset files:", path)

print(os.listdir(path))
file_path = os.path.join(path, "youtube_shorts_dataset.csv")
df = pd.read_csv(file_path)

df = df.drop_duplicates()
df = df.dropna()

df
```
