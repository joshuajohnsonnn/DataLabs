```python
import kagglehub
import pandas as pd
import os

#Download latest version
path = kagglehub.dataset_download("dianatofficial/data-engineering-ai-pipelines")

print("Path to dataset files:", path)
#see whats inside the path
print(os.listdir(path))

#load the csv file into a pandas dataframe
file_path = os.path.join(path, "data_engineering_ai_pipelines.csv")
df = pd.read_csv(file_path)

#clean the dataframe by dropping any rows with missing values
df = df.drop_duplicates()
df = df.dropna()

print(df.head())

df.info()

print(f"Rows before cleaning: {len(df):,}")
print(f"Duplicate rows: {df.duplicated().sum():,}")
print(f"Missing values: {df.isna().sum().sum():,}")
print(f"Rows after cleaning: {len(df):,}")

df.describe(include='all')
```
